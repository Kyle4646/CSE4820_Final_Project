```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import time

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input, Conv1D, MaxPooling1D,
    Flatten, Dense, Dropout
)
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam
from tensorflow.keras import backend as K

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    ParameterSampler
)
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score, f1_score

# ============================================
# 1. FEATURES AND TARGET
# ============================================

significant_features = [
    "VC_windspeed",
    "VC_cloudcover",
    "VC_visibility",
    "VC_temp",
    "VC_humidity",
    "VC_precip",
    "VC_2h_temp",
    "VC_2h_humidity",
    "VC_2h_precip",
    "VC_2h_windspeed",
    "VC_2h_visibility",
    "Market Share",
    "Passenger Load Factor",
    "OTP Index",
    "Hourly_Airport_Departures",
    "Distance",
    "Airport_Congestion_Index"
]

X = data[significant_features].apply(
    pd.to_numeric, errors="coerce"
)

y = (data["Departure Delay"] > 0).astype(int)

# Hold out 20% for final testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# ============================================
# 2. CNN HYPERPARAMETER SEARCH SPACE
# ============================================

param_grid = {
    "filters": [
        (16,),
        (32,),
        (64,),
        (128,),
        (16, 32),
        (32, 64),
        (64, 32),
        (64, 128),
        (32, 64, 128),
        (64, 32, 16)
    ],

    "kernel_size": [2, 3, 5],

    "pooling": [True, False],

    "dense_units": [16, 32, 64, 128],

    "dropout": [0.0, 0.2, 0.3, 0.5],

    "l2_strength": [0.0, 0.0001, 0.001],

    "learning_rate": [0.001, 0.0005]
}

N_TRIALS = 20
N_FOLDS = 3
MAX_EPOCHS = 50
BATCH_SIZE = 64

configurations = list(ParameterSampler(
    param_grid,
    n_iter=N_TRIALS,
    random_state=42
))

print(f"Architectures: {N_TRIALS}")
print(f"CV Folds: {N_FOLDS}")
print(f"Total CNN Fits: {N_TRIALS * N_FOLDS}")

# ============================================
# 3. DYNAMIC CNN ARCHITECTURE
# ============================================

def build_cnn(n_features, params):

    model = Sequential()

    model.add(Input(shape=(n_features, 1)))

    # Dynamically create convolutional layers
    for filters in params["filters"]:

        model.add(
            Conv1D(
                filters=filters,
                kernel_size=params["kernel_size"],
                activation="relu",
                padding="same",
                kernel_regularizer=l2(
                    params["l2_strength"]
                )
            )
        )

        # Optional max pooling
        if params["pooling"]:

            model.add(
                MaxPooling1D(
                    pool_size=2,
                    padding="same"
                )
            )

    model.add(Flatten())

    model.add(
        Dense(
            params["dense_units"],
            activation="relu",
            kernel_regularizer=l2(
                params["l2_strength"]
            )
        )
    )

    model.add(Dropout(params["dropout"]))

    model.add(Dense(1, activation="sigmoid"))

    model.compile(
        optimizer=Adam(
            learning_rate=params["learning_rate"]
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model

# ============================================
# 4. PREPARE 3 CROSS-VALIDATION FOLDS
# ============================================

cv = StratifiedKFold(
    n_splits=N_FOLDS,
    shuffle=True,
    random_state=42
)

prepared_folds = []

# Preprocess each fold once to save time

for train_idx, val_idx in cv.split(X_train, y_train):

    X_tr = X_train.iloc[train_idx]
    X_va = X_train.iloc[val_idx]

    y_tr = y_train.iloc[train_idx].to_numpy()
    y_va = y_train.iloc[val_idx].to_numpy()

    # Imputation
    imputer = SimpleImputer(
        strategy="median",
        keep_empty_features=True
    )

    X_tr = imputer.fit_transform(X_tr)
    X_va = imputer.transform(X_va)

    # Scaling
    scaler = StandardScaler()

    X_tr = scaler.fit_transform(X_tr)
    X_va = scaler.transform(X_va)

    # Convert to CNN input shape
    X_tr = X_tr.astype(np.float32)[..., np.newaxis]
    X_va = X_va.astype(np.float32)[..., np.newaxis]

    prepared_folds.append(
        (X_tr, X_va, y_tr, y_va)
    )

# ============================================
# 5. ARCHITECTURE SEARCH
# ============================================

results = []

best_f1 = -1
best_params = None

total_start = time.time()

for trial, params in enumerate(
    configurations, start=1
):

    trial_start = time.time()

    print(
        f"\n========== TRIAL {trial}/{N_TRIALS} ==========",
        flush=True
    )

    print(params, flush=True)

    train_f1_scores = []
    val_f1_scores = []

    train_acc_scores = []
    val_acc_scores = []

    epoch_counts = []

    for fold, (X_tr, X_va, y_tr, y_va) in enumerate(
        prepared_folds, start=1
    ):

        fold_start = time.time()

        K.clear_session()

        tf.keras.utils.set_random_seed(42 + fold)

        model = build_cnn(
            len(significant_features),
            params
        )

        early_stop = EarlyStopping(
            monitor="val_loss",
            patience=5,
            restore_best_weights=True
        )

        # Train CNN
        history = model.fit(
            X_tr,
            y_tr,
            validation_data=(X_va, y_va),
            epochs=MAX_EPOCHS,
            batch_size=BATCH_SIZE,
            callbacks=[early_stop],
            verbose=0
        )

        # Training predictions
        train_prob = model.predict(
            X_tr,
            verbose=0
        ).ravel()

        # Validation predictions
        val_prob = model.predict(
            X_va,
            verbose=0
        ).ravel()

        train_pred = (train_prob >= 0.5).astype(int)
        val_pred = (val_prob >= 0.5).astype(int)

        # F1 scores
        train_f1 = f1_score(
            y_tr,
            train_pred,
            zero_division=0
        )

        val_f1 = f1_score(
            y_va,
            val_pred,
            zero_division=0
        )

        # Accuracy scores
        train_acc = accuracy_score(
            y_tr, train_pred
        )

        val_acc = accuracy_score(
            y_va, val_pred
        )

        train_f1_scores.append(train_f1)
        val_f1_scores.append(val_f1)

        train_acc_scores.append(train_acc)
        val_acc_scores.append(val_acc)

        epoch_counts.append(
            len(history.history["loss"])
        )

        fold_time = time.time() - fold_start

        print(
            f"Fold {fold}/{N_FOLDS} | "
            f"Train F1: {train_f1:.4f} | "
            f"Val F1: {val_f1:.4f} | "
            f"Val Acc: {val_acc:.4f} | "
            f"Time: {fold_time:.1f}s",
            flush=True
        )

    # ========================================
    # ARCHITECTURE RESULTS
    # ========================================

    mean_train_f1 = np.mean(train_f1_scores)
    mean_val_f1 = np.mean(val_f1_scores)

    mean_train_acc = np.mean(train_acc_scores)
    mean_val_acc = np.mean(val_acc_scores)

    trial_time = time.time() - trial_start

    results.append({
        "Trial": trial,
        **params,
        "Conv Layers": len(params["filters"]),
        "Trainable Parameters": model.count_params(),
        "Train F1": mean_train_f1,
        "Validation F1": mean_val_f1,
        "Train Accuracy": mean_train_acc,
        "Validation Accuracy": mean_val_acc,
        "F1 Std": np.std(val_f1_scores),
        "F1 Gap": mean_train_f1 - mean_val_f1,
        "Accuracy Gap": mean_train_acc - mean_val_acc,
        "Mean Epochs": np.mean(epoch_counts),
        "Time (seconds)": trial_time
    })

    # Update best architecture
    if mean_val_f1 > best_f1:

        best_f1 = mean_val_f1
        best_params = params.copy()

        print("NEW BEST ARCHITECTURE!", flush=True)

    print(f"Mean Training F1: {mean_train_f1:.4f}")
    print(f"Mean Validation F1: {mean_val_f1:.4f}")
    print(f"Mean Validation Accuracy: {mean_val_acc:.4f}")
    print(f"Best F1 so far: {best_f1:.4f}")
    print(f"Trial time: {trial_time / 60:.2f} minutes")

    # Estimate remaining runtime
    elapsed = time.time() - total_start

    estimated_remaining = (
        elapsed / trial
    ) * (N_TRIALS - trial)

    print(
        f"Estimated remaining: "
        f"{estimated_remaining / 60:.1f} minutes",
        flush=True
    )

    # Save results after every completed trial
    pd.DataFrame(results).to_csv(
        "CNN_architecture_search_3fold.csv",
        index=False
    )

# ============================================
# 6. RANK ALL ARCHITECTURES
# ============================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    "Validation F1",
    ascending=False
).reset_index(drop=True)

print("\n========== TOP 10 ARCHITECTURES ==========")

display(results_df.head(10).round(4))

print("\n========== BEST PARAMETERS ==========")

for key, value in best_params.items():
    print(f"{key}: {value}")

print(f"\nBest Validation F1: {best_f1:.4f}")

print(
    f"Best Architecture Accuracy: "
    f"{results_df.iloc[0]['Validation Accuracy']:.4f}"
)

print(
    f"Total runtime: "
    f"{(time.time() - total_start) / 60:.2f} minutes"
)

# ============================================
# 7. BEST ARCHITECTURE BY ACCURACY
# ============================================

accuracy_ranked = results_df.sort_values(
    "Validation Accuracy",
    ascending=False
)

print("\n========== TOP 5 BY ACCURACY ==========")

display(accuracy_ranked.head(5).round(4))

# ============================================
# 8. BEST CNN ARCHITECTURE SUMMARY
# ============================================

K.clear_session()

best_model = build_cnn(
    len(significant_features),
    best_params
)

best_model.summary()

# ============================================
# 9. PLOT TOP 10 ARCHITECTURES BY F1
# ============================================

top_10 = results_df.head(10)

positions = np.arange(len(top_10))

plt.figure(figsize=(12, 6))

plt.bar(
    positions - 0.2,
    top_10["Train F1"],
    width=0.4,
    label="Training F1"
)

plt.bar(
    positions + 0.2,
    top_10["Validation F1"],
    width=0.4,
    label="Validation F1"
)

plt.xticks(
    positions,
    [f"Trial {t}" for t in top_10["Trial"]]
)

plt.ylabel("F1 Score")
plt.xlabel("CNN Architecture")
plt.title("Top 10 CNN Architectures: 3-Fold CV")
plt.ylim(0, 1)
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()

# ============================================
# 10. PLOT TRAINING VS VALIDATION ACCURACY
# ============================================

plt.figure(figsize=(12, 6))

plt.bar(
    positions - 0.2,
    top_10["Train Accuracy"],
    width=0.4,
    label="Training Accuracy"
)

plt.bar(
    positions + 0.2,
    top_10["Validation Accuracy"],
    width=0.4,
    label="Validation Accuracy"
)

plt.xticks(
    positions,
    [f"Trial {t}" for t in top_10["Trial"]]
)

plt.ylabel("Accuracy")
plt.xlabel("CNN Architecture")
plt.title("CNN Training vs Validation Accuracy")
plt.ylim(0, 1)
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input, Conv1D, MaxPooling1D,
    GlobalAveragePooling1D, Dense, Dropout
)
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam
from tensorflow.keras import backend as K

from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score, f1_score,
    precision_score, recall_score,
    confusion_matrix, ConfusionMatrixDisplay
)

# ============================================
# 1. FEATURES AND TARGET
# ============================================

significant_features = [
    "VC_windspeed",
    "VC_cloudcover",
    "VC_visibility",
    "VC_temp",
    "VC_humidity",
    "VC_precip",
    "VC_2h_temp",
    "VC_2h_humidity",
    "VC_2h_precip",
    "VC_2h_windspeed",
    "VC_2h_visibility",
    "Market Share",
    "Passenger Load Factor",
    "OTP Index",
    "Hourly_Airport_Departures",
    "Distance",
    "Airport_Congestion_Index"
]

X = data[significant_features].apply(
    pd.to_numeric, errors="coerce"
)

y = (data["Departure Delay"] > 0).astype(int)

# Same 80/20 split as previous models
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# ============================================
# 2. NEW CNN ARCHITECTURE
# ============================================

def build_cnn(n_features):

    model = Sequential([
        Input(shape=(n_features, 1)),

        Conv1D(
            filters=32,
            kernel_size=3,
            activation="relu",
            padding="same",
            kernel_regularizer=l2(0.0001)
        ),

        MaxPooling1D(pool_size=2),

        Conv1D(
            filters=16,
            kernel_size=3,
            activation="relu",
            padding="same",
            kernel_regularizer=l2(0.0001)
        ),

        GlobalAveragePooling1D(),

        Dense(
            32,
            activation="relu",
            kernel_regularizer=l2(0.0001)
        ),

        Dropout(0.3),

        Dense(1, activation="sigmoid")
    ])

    model.compile(
        optimizer=Adam(learning_rate=0.0005),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


preview = build_cnn(len(significant_features))
preview.summary()

# ============================================
# 3. STRATIFIED 5-FOLD CV
# ============================================

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

fold_train_f1 = []
fold_val_f1 = []
fold_train_acc = []
fold_val_acc = []
fold_precision = []
fold_recall = []
histories = []

for fold, (train_idx, val_idx) in enumerate(
    cv.split(X_train, y_train), start=1
):

    print(f"\n========== FOLD {fold}/5 ==========")

    X_fold_train = X_train.iloc[train_idx]
    X_fold_val = X_train.iloc[val_idx]

    y_fold_train = y_train.iloc[train_idx]
    y_fold_val = y_train.iloc[val_idx]

    # ----------------------------------------
    # Imputation
    # ----------------------------------------

    imputer = SimpleImputer(
        strategy="median",
        keep_empty_features=True
    )

    X_fold_train = imputer.fit_transform(X_fold_train)
    X_fold_val = imputer.transform(X_fold_val)

    # ----------------------------------------
    # Standardization
    # ----------------------------------------

    scaler = StandardScaler()

    X_fold_train = scaler.fit_transform(X_fold_train)
    X_fold_val = scaler.transform(X_fold_val)

    # ----------------------------------------
    # Reshape for CNN
    # ----------------------------------------

    X_fold_train = X_fold_train.reshape(
        X_fold_train.shape[0],
        X_fold_train.shape[1],
        1
    )

    X_fold_val = X_fold_val.reshape(
        X_fold_val.shape[0],
        X_fold_val.shape[1],
        1
    )

    # ----------------------------------------
    # Train fresh model
    # ----------------------------------------

    K.clear_session()
    tf.keras.utils.set_random_seed(42 + fold)

    model = build_cnn(X_fold_train.shape[1])

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True
    )

    history = model.fit(
        X_fold_train,
        y_fold_train,
        validation_data=(X_fold_val, y_fold_val),
        epochs=100,
        batch_size=32,
        callbacks=[early_stop],
        verbose=0
    )

    histories.append(history.history)

    # ----------------------------------------
    # Predictions
    # ----------------------------------------

    train_prob = model.predict(
        X_fold_train, verbose=0
    ).ravel()

    val_prob = model.predict(
        X_fold_val, verbose=0
    ).ravel()

    train_pred = (train_prob >= 0.5).astype(int)
    val_pred = (val_prob >= 0.5).astype(int)

    # ----------------------------------------
    # Metrics
    # ----------------------------------------

    train_f1 = f1_score(y_fold_train, train_pred)
    val_f1 = f1_score(y_fold_val, val_pred)

    train_acc = accuracy_score(y_fold_train, train_pred)
    val_acc = accuracy_score(y_fold_val, val_pred)

    precision = precision_score(
        y_fold_val, val_pred, zero_division=0
    )

    recall = recall_score(
        y_fold_val, val_pred, zero_division=0
    )

    fold_train_f1.append(train_f1)
    fold_val_f1.append(val_f1)

    fold_train_acc.append(train_acc)
    fold_val_acc.append(val_acc)

    fold_precision.append(precision)
    fold_recall.append(recall)

    print(f"Epochs:        {len(history.history['loss'])}")
    print(f"Training F1:   {train_f1:.4f}")
    print(f"Validation F1: {val_f1:.4f}")
    print(f"Validation Acc:{val_acc:.4f}")

# ============================================
# 4. CROSS-VALIDATION RESULTS
# ============================================

results = pd.DataFrame({
    "Fold": range(1, 6),
    "Training Accuracy": fold_train_acc,
    "Validation Accuracy": fold_val_acc,
    "Training F1": fold_train_f1,
    "Validation F1": fold_val_f1,
    "Precision": fold_precision,
    "Recall": fold_recall
})

print("\n========== 5-FOLD RESULTS ==========")
display(results.round(4))

mean_train_f1 = np.mean(fold_train_f1)
mean_val_f1 = np.mean(fold_val_f1)

print("\n========== AVERAGES ==========")

print(f"Mean Training Accuracy:   {np.mean(fold_train_acc):.4f}")
print(f"Mean Validation Accuracy: {np.mean(fold_val_acc):.4f}")

print(f"Mean Training F1:         {mean_train_f1:.4f}")
print(f"Mean Validation F1:       {mean_val_f1:.4f}")

print(f"Mean Precision:           {np.mean(fold_precision):.4f}")
print(f"Mean Recall:              {np.mean(fold_recall):.4f}")

print(f"F1 Standard Deviation:    {np.std(fold_val_f1):.4f}")
print(f"F1 Generalization Gap:    {mean_train_f1 - mean_val_f1:.4f}")

# ============================================
# 5. TRAINING VS VALIDATION LOSS
# ============================================

plt.figure(figsize=(10, 6))

for h in histories:

    epochs = np.arange(1, len(h["loss"]) + 1)

    plt.plot(
        epochs,
        h["loss"],
        color="blue",
        alpha=0.15
    )

    plt.plot(
        epochs,
        h["val_loss"],
        color="orange",
        alpha=0.15
    )

max_epochs = max(len(h["loss"]) for h in histories)

mean_train_loss = []
mean_val_loss = []

for epoch in range(max_epochs):

    train_values = [
        h["loss"][epoch]
        for h in histories
        if len(h["loss"]) > epoch
    ]

    val_values = [
        h["val_loss"][epoch]
        for h in histories
        if len(h["val_loss"]) > epoch
    ]

    mean_train_loss.append(np.mean(train_values))
    mean_val_loss.append(np.mean(val_values))

plt.plot(
    range(1, max_epochs + 1),
    mean_train_loss,
    color="blue",
    linewidth=2.5,
    label="Mean Training Loss"
)

plt.plot(
    range(1, max_epochs + 1),
    mean_val_loss,
    color="orange",
    linewidth=2.5,
    label="Mean Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss (Binary Cross-Entropy + L2)")
plt.title("CNN: 5-Fold Training vs Validation Loss")
plt.legend()
plt.grid(True)
plt.show()

# ============================================
# 6. TRAINING VS VALIDATION F1
# ============================================

plt.figure(figsize=(7, 5))

plt.bar(
    ["Training F1", "Validation F1"],
    [mean_train_f1, mean_val_f1]
)

plt.ylabel("F1 Score")
plt.title("CNN Training vs Validation F1")
plt.ylim(0, 1)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import time

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input, Conv1D, MaxPooling1D,
    Flatten, Dense, Dropout
)
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam
from tensorflow.keras import backend as K

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    ParameterSampler
)
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score, f1_score

# ============================================
# 1. FEATURES AND TARGET
# ============================================

significant_features = [
    "VC_windspeed",
    "VC_cloudcover",
    "VC_visibility",
    "VC_temp",
    "VC_humidity",
    "VC_precip",
    "VC_2h_temp",
    "VC_2h_humidity",
    "VC_2h_precip",
    "VC_2h_windspeed",
    "VC_2h_visibility",
    "Market Share",
    "Passenger Load Factor",
    "OTP Index",
    "Hourly_Airport_Departures",
    "Distance",
    "Airport_Congestion_Index"
]

X = data[significant_features].apply(
    pd.to_numeric, errors="coerce"
)

y = (data["Departure Delay"] > 0).astype(int)

# Hold out 20% for final testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# ============================================
# 2. DEEP CNN SEARCH SPACE
# ============================================

param_grid = {
    "filters": [
        # 1 convolutional layer
        (32,),
        (64,),
        (128,),

        # 2 convolutional layers
        (32, 64),
        (64, 128),

        # 3 convolutional layers
        (32, 64, 128),
        (64, 128, 256),

        # 4 convolutional layers
        (16, 32, 64, 128),
        (32, 64, 128, 256),
        (64, 128, 128, 64),

        # 5 convolutional layers
        (16, 32, 64, 128, 256),
        (32, 64, 128, 128, 64),
        (64, 128, 256, 128, 64),

        # 6 convolutional layers
        (16, 32, 64, 128, 128, 256),
        (32, 64, 128, 256, 128, 64),
        (64, 128, 256, 256, 128, 64)
    ],

    "kernel_size": [2, 3, 5],

    "pooling": [True, False],

    "dense_units": [32, 64, 128, 256],

    "dropout": [0.2, 0.3, 0.4, 0.5],

    "l2_strength": [0.0, 0.0001, 0.001],

    "learning_rate": [0.001, 0.0005]
}

N_TRIALS = 20
N_FOLDS = 3
MAX_EPOCHS = 50
BATCH_SIZE = 64

# ============================================
# 3. GENERATE ARCHITECTURES
# ============================================

configurations = []

# Guarantee every depth is represented.
# More trials are allocated to deeper models.

trials_per_depth = {
    1: 2,
    2: 2,
    3: 3,
    4: 4,
    5: 4,
    6: 5
}

for depth, n_trials in trials_per_depth.items():

    depth_grid = param_grid.copy()

    depth_grid["filters"] = [
        filters
        for filters in param_grid["filters"]
        if len(filters) == depth
    ]

    sampled = list(ParameterSampler(
        depth_grid,
        n_iter=n_trials,
        random_state=42 + depth
    ))

    configurations.extend(sampled)

N_TRIALS = len(configurations)

print(f"Architectures: {N_TRIALS}")
print(f"CV Folds: {N_FOLDS}")
print(f"Total CNN Fits: {N_TRIALS * N_FOLDS}")

# ============================================
# 4. DYNAMIC DEEP CNN ARCHITECTURE
# ============================================

def build_cnn(n_features, params):

    model = Sequential()

    model.add(Input(shape=(n_features, 1)))

    # Automatically create convolutional layers
    for i, filters in enumerate(params["filters"]):

        model.add(
            Conv1D(
                filters=filters,
                kernel_size=params["kernel_size"],
                activation="relu",
                padding="same",
                kernel_regularizer=l2(
                    params["l2_strength"]
                )
            )
        )

        # Pool after every second convolution.
        # For a single-layer CNN, pool after that layer.
        if params["pooling"] and (
            (i + 1) % 2 == 0
            or len(params["filters"]) == 1
        ):

            model.add(
                MaxPooling1D(
                    pool_size=2,
                    padding="same"
                )
            )

    # Convert feature maps into a vector
    model.add(Flatten())

    model.add(
        Dense(
            params["dense_units"],
            activation="relu",
            kernel_regularizer=l2(
                params["l2_strength"]
            )
        )
    )

    model.add(Dropout(params["dropout"]))

    model.add(Dense(1, activation="sigmoid"))

    model.compile(
        optimizer=Adam(
            learning_rate=params["learning_rate"]
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model

# ============================================
# 5. PREPARE CROSS-VALIDATION FOLDS
# ============================================

cv = StratifiedKFold(
    n_splits=N_FOLDS,
    shuffle=True,
    random_state=42
)

prepared_folds = []

# Preprocess each fold once to save time

for train_idx, val_idx in cv.split(X_train, y_train):

    X_tr = X_train.iloc[train_idx]
    X_va = X_train.iloc[val_idx]

    y_tr = y_train.iloc[train_idx].to_numpy()
    y_va = y_train.iloc[val_idx].to_numpy()

    # Imputation
    imputer = SimpleImputer(
        strategy="median",
        keep_empty_features=True
    )

    X_tr = imputer.fit_transform(X_tr)
    X_va = imputer.transform(X_va)

    # Standardization
    scaler = StandardScaler()

    X_tr = scaler.fit_transform(X_tr)
    X_va = scaler.transform(X_va)

    # CNN input shape
    X_tr = X_tr.astype(np.float32)[..., np.newaxis]
    X_va = X_va.astype(np.float32)[..., np.newaxis]

    prepared_folds.append(
        (X_tr, X_va, y_tr, y_va)
    )

# ============================================
# 6. ARCHITECTURE SEARCH
# ============================================

results = []

best_f1 = -1
best_params = None

total_start = time.time()

try:

    for trial, params in enumerate(
        configurations, start=1
    ):

        trial_start = time.time()

        print(
            f"\n========== TRIAL {trial}/{N_TRIALS} ==========",
            flush=True
        )

        print(
            f"Convolutional Layers: {len(params['filters'])}"
        )

        print(params, flush=True)

        train_f1_scores = []
        val_f1_scores = []

        train_acc_scores = []
        val_acc_scores = []

        epoch_counts = []

        for fold, (X_tr, X_va, y_tr, y_va) in enumerate(
            prepared_folds, start=1
        ):

            fold_start = time.time()

            K.clear_session()

            tf.keras.utils.set_random_seed(42 + fold)

            model = build_cnn(
                len(significant_features),
                params
            )

            early_stop = EarlyStopping(
                monitor="val_loss",
                patience=5,
                restore_best_weights=True
            )

            print(
                f"Training fold {fold}/{N_FOLDS}...",
                flush=True
            )

            # Train CNN
            history = model.fit(
                X_tr,
                y_tr,
                validation_data=(X_va, y_va),
                epochs=MAX_EPOCHS,
                batch_size=BATCH_SIZE,
                callbacks=[early_stop],
                verbose=0
            )

            # Training predictions
            train_prob = model.predict(
                X_tr,
                verbose=0
            ).ravel()

            # Validation predictions
            val_prob = model.predict(
                X_va,
                verbose=0
            ).ravel()

            train_pred = (train_prob >= 0.5).astype(int)
            val_pred = (val_prob >= 0.5).astype(int)

            # F1 scores
            train_f1 = f1_score(
                y_tr,
                train_pred,
                zero_division=0
            )

            val_f1 = f1_score(
                y_va,
                val_pred,
                zero_division=0
            )

            # Accuracy scores
            train_acc = accuracy_score(
                y_tr, train_pred
            )

            val_acc = accuracy_score(
                y_va, val_pred
            )

            train_f1_scores.append(train_f1)
            val_f1_scores.append(val_f1)

            train_acc_scores.append(train_acc)
            val_acc_scores.append(val_acc)

            epoch_counts.append(
                len(history.history["loss"])
            )

            fold_time = time.time() - fold_start

            print(
                f"Fold {fold}/{N_FOLDS} | "
                f"Train F1: {train_f1:.4f} | "
                f"Val F1: {val_f1:.4f} | "
                f"Val Acc: {val_acc:.4f} | "
                f"Epochs: {epoch_counts[-1]} | "
                f"Time: {fold_time:.1f}s",
                flush=True
            )

        # ====================================
        # ARCHITECTURE RESULTS
        # ====================================

        mean_train_f1 = np.mean(train_f1_scores)
        mean_val_f1 = np.mean(val_f1_scores)

        mean_train_acc = np.mean(train_acc_scores)
        mean_val_acc = np.mean(val_acc_scores)

        trial_time = time.time() - trial_start

        results.append({
            "Trial": trial,
            **params,
            "Conv Layers": len(params["filters"]),
            "Trainable Parameters": model.count_params(),
            "Train F1": mean_train_f1,
            "Validation F1": mean_val_f1,
            "Train Accuracy": mean_train_acc,
            "Validation Accuracy": mean_val_acc,
            "F1 Std": np.std(val_f1_scores),
            "F1 Gap": mean_train_f1 - mean_val_f1,
            "Accuracy Gap": mean_train_acc - mean_val_acc,
            "Mean Epochs": np.mean(epoch_counts),
            "Time (seconds)": trial_time
        })

        # Update best architecture
        if mean_val_f1 > best_f1:

            best_f1 = mean_val_f1
            best_params = params.copy()

            print("NEW BEST ARCHITECTURE!", flush=True)

        print(f"Mean Training F1: {mean_train_f1:.4f}")
        print(f"Mean Validation F1: {mean_val_f1:.4f}")
        print(f"Mean Validation Accuracy: {mean_val_acc:.4f}")
        print(f"Best F1 so far: {best_f1:.4f}")

        print(
            f"Trial time: {trial_time / 60:.2f} minutes"
        )

        elapsed = time.time() - total_start

        estimated_remaining = (
            elapsed / trial
        ) * (N_TRIALS - trial)

        print(
            f"Estimated remaining: "
            f"{estimated_remaining / 60:.1f} minutes",
            flush=True
        )

        # Save after every completed architecture
        pd.DataFrame(results).to_csv(
            "Deep_CNN_Search_3fold.csv",
            index=False
        )

except KeyboardInterrupt:

    print(
        "\nSearch interrupted. "
        "Completed architectures have been preserved."
    )

# ============================================
# 7. RANK ARCHITECTURES
# ============================================

if not results:
    raise RuntimeError("No architectures completed.")

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    "Validation F1",
    ascending=False
).reset_index(drop=True)

print("\n========== TOP 10 ARCHITECTURES ==========")

display(results_df.head(10).round(4))

print("\n========== BEST PARAMETERS ==========")

for key, value in best_params.items():
    print(f"{key}: {value}")

print(f"\nBest Validation F1: {best_f1:.4f}")

print(
    f"Best Architecture Accuracy: "
    f"{results_df.iloc[0]['Validation Accuracy']:.4f}"
)

print(
    f"Total runtime: "
    f"{(time.time() - total_start) / 60:.2f} minutes"
)

# ============================================
# 8. BEST ARCHITECTURES BY ACCURACY
# ============================================

accuracy_ranked = results_df.sort_values(
    "Validation Accuracy",
    ascending=False
)

print("\n========== TOP 5 BY ACCURACY ==========")

display(accuracy_ranked.head(5).round(4))

# ============================================
# 9. COMPARE PERFORMANCE BY CNN DEPTH
# ============================================

depth_results = results_df.groupby(
    "Conv Layers"
).agg(
    Architectures=("Trial", "count"),
    Mean_Train_F1=("Train F1", "mean"),
    Mean_Validation_F1=("Validation F1", "mean"),
    Best_Validation_F1=("Validation F1", "max"),
    Mean_Accuracy=("Validation Accuracy", "mean"),
    Best_Accuracy=("Validation Accuracy", "max"),
    Mean_F1_Gap=("F1 Gap", "mean"),
    Mean_Parameters=("Trainable Parameters", "mean")
)

print("\n========== PERFORMANCE BY DEPTH ==========")

display(depth_results.round(4))

# ============================================
# 10. BEST MODEL SUMMARY
# ============================================

K.clear_session()

best_model = build_cnn(
    len(significant_features),
    best_params
)

best_model.summary()

# ============================================
# 11. TOP 10 ARCHITECTURES BY F1
# ============================================

top_10 = results_df.head(10)

positions = np.arange(len(top_10))

plt.figure(figsize=(12, 6))

plt.bar(
    positions - 0.2,
    top_10["Train F1"],
    width=0.4,
    label="Training F1"
)

plt.bar(
    positions + 0.2,
    top_10["Validation F1"],
    width=0.4,
    label="Validation F1"
)

plt.xticks(
    positions,
    [
        f"Trial {t}\n{d} Conv"
        for t, d in zip(
            top_10["Trial"],
            top_10["Conv Layers"]
        )
    ]
)

plt.ylabel("F1 Score")
plt.xlabel("CNN Architecture")
plt.title("Top 10 Deep CNN Architectures: 3-Fold CV")
plt.ylim(0, 1)
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()

# ============================================
# 12. F1 VS NETWORK DEPTH
# ============================================

plt.figure(figsize=(9, 5))

plt.plot(
    depth_results.index,
    depth_results["Mean_Train_F1"],
    marker="o",
    label="Mean Training F1"
)

plt.plot(
    depth_results.index,
    depth_results["Mean_Validation_F1"],
    marker="o",
    label="Mean Validation F1"
)

plt.plot(
    depth_results.index,
    depth_results["Best_Validation_F1"],
    marker="o",
    linestyle="--",
    label="Best Validation F1"
)

plt.xlabel("Number of Convolutional Layers")
plt.ylabel("F1 Score")
plt.title("Effect of CNN Depth on Performance")
plt.xticks(depth_results.index)
plt.legend()
plt.grid(True)
plt.show()

# ============================================
# 13. ACCURACY VS NETWORK DEPTH
# ============================================

plt.figure(figsize=(9, 5))

plt.plot(
    depth_results.index,
    depth_results["Mean_Accuracy"],
    marker="o",
    label="Mean Validation Accuracy"
)

plt.plot(
    depth_results.index,
    depth_results["Best_Accuracy"],
    marker="o",
    linestyle="--",
    label="Best Validation Accuracy"
)

plt.xlabel("Number of Convolutional Layers")
plt.ylabel("Accuracy")
plt.title("Effect of CNN Depth on Accuracy")
plt.xticks(depth_results.index)
plt.legend()
plt.grid(True)
plt.show()

# ============================================
# 14. MODEL SIZE VS VALIDATION F1
# ============================================

plt.figure(figsize=(9, 5))

scatter = plt.scatter(
    results_df["Trainable Parameters"],
    results_df["Validation F1"],
    c=results_df["Conv Layers"],
    cmap="viridis",
    s=80
)

plt.colorbar(
    scatter,
    label="Convolutional Layers"
)

plt.xscale("log")
plt.xlabel("Trainable Parameters (Log Scale)")
plt.ylabel("Validation F1")
plt.title("CNN Model Complexity vs Performance")
plt.grid(True)
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import time

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input, Conv1D, Flatten, Dense, Dropout
)
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam
from tensorflow.keras import backend as K

from sklearn.model_selection import (
    train_test_split, StratifiedKFold
)
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score, f1_score

# ============================================
# 1. FEATURES
# ============================================

fixed_features = [
    "Market Share",
    "Passenger Load Factor",
    "OTP Index",
    "Hourly_Airport_Departures",
    "Distance",
    "Airport_Congestion_Index"
]

weather_features = [
    "VC_windspeed",
    "VC_cloudcover",
    "VC_visibility",
    "VC_temp",
    "VC_humidity",
    "VC_precip",
    "VC_2h_temp",
    "VC_2h_humidity",
    "VC_2h_precip",
    "VC_2h_windspeed",
    "VC_2h_visibility"
]

# Preserve original CNN feature ordering
all_features = weather_features + fixed_features

X = data[all_features].apply(
    pd.to_numeric, errors="coerce"
)

y = (data["Departure Delay"] > 0).astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

N_FOLDS = 3
MAX_EPOCHS = 50
BATCH_SIZE = 64

# Stop when no removal improves F1
STOP_EARLY = True

# ============================================
# 2. WINNING CNN ARCHITECTURE
# ============================================

def build_cnn(n_features):

    model = Sequential([
        Input(shape=(n_features, 1)),

        Conv1D(
            filters=64,
            kernel_size=5,
            activation="relu",
            padding="same",
            kernel_regularizer=l2(0.0001)
        ),

        Conv1D(
            filters=128,
            kernel_size=5,
            activation="relu",
            padding="same",
            kernel_regularizer=l2(0.0001)
        ),

        Flatten(),

        Dense(
            64,
            activation="relu",
            kernel_regularizer=l2(0.0001)
        ),

        Dropout(0.5),

        Dense(1, activation="sigmoid")
    ])

    model.compile(
        optimizer=Adam(learning_rate=0.0005),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model

build_cnn(len(all_features)).summary()

# ============================================
# 3. CROSS-VALIDATION
# ============================================

cv = StratifiedKFold(
    n_splits=N_FOLDS,
    shuffle=True,
    random_state=42
)

splits = list(cv.split(X_train, y_train))

def evaluate_features(features):

    train_f1_scores = []
    val_f1_scores = []

    train_acc_scores = []
    val_acc_scores = []

    start = time.time()

    for fold, (train_idx, val_idx) in enumerate(
        splits, start=1
    ):

        X_tr = X_train.iloc[train_idx][features]
        X_va = X_train.iloc[val_idx][features]

        y_tr = y_train.iloc[train_idx].to_numpy()
        y_va = y_train.iloc[val_idx].to_numpy()

        # Imputation
        imputer = SimpleImputer(
            strategy="median",
            keep_empty_features=True
        )

        X_tr = imputer.fit_transform(X_tr)
        X_va = imputer.transform(X_va)

        # Scaling
        scaler = StandardScaler()

        X_tr = scaler.fit_transform(X_tr)
        X_va = scaler.transform(X_va)

        # CNN input shape
        X_tr = X_tr.astype(np.float32)[..., None]
        X_va = X_va.astype(np.float32)[..., None]

        K.clear_session()

        tf.keras.utils.set_random_seed(42 + fold)

        model = build_cnn(len(features))

        early_stop = EarlyStopping(
            monitor="val_loss",
            patience=5,
            restore_best_weights=True
        )

        history = model.fit(
            X_tr,
            y_tr,
            validation_data=(X_va, y_va),
            epochs=MAX_EPOCHS,
            batch_size=BATCH_SIZE,
            callbacks=[early_stop],
            verbose=0
        )

        train_prob = model.predict(
            X_tr, verbose=0
        ).ravel()

        val_prob = model.predict(
            X_va, verbose=0
        ).ravel()

        train_pred = (train_prob >= 0.5).astype(int)
        val_pred = (val_prob >= 0.5).astype(int)

        train_f1 = f1_score(
            y_tr, train_pred, zero_division=0
        )

        val_f1 = f1_score(
            y_va, val_pred, zero_division=0
        )

        train_f1_scores.append(train_f1)
        val_f1_scores.append(val_f1)

        train_acc_scores.append(
            accuracy_score(y_tr, train_pred)
        )

        val_acc_scores.append(
            accuracy_score(y_va, val_pred)
        )

        print(
            f"  Fold {fold}/{N_FOLDS} | "
            f"Train F1: {train_f1:.4f} | "
            f"Val F1: {val_f1:.4f}",
            flush=True
        )

    return {
        "Features": features.copy(),
        "Number of Features": len(features),
        "Train F1": np.mean(train_f1_scores),
        "Validation F1": np.mean(val_f1_scores),
        "Train Accuracy": np.mean(train_acc_scores),
        "Validation Accuracy": np.mean(val_acc_scores),
        "F1 Std": np.std(val_f1_scores),
        "F1 Gap": (
            np.mean(train_f1_scores)
            - np.mean(val_f1_scores)
        ),
        "Trainable Parameters": model.count_params(),
        "Time (seconds)": time.time() - start
    }

# ============================================
# 4. BASELINE: ALL 17 FEATURES
# ============================================

results = []
selection_path = []

current_features = all_features.copy()

total_start = time.time()

print("\n========== ALL FEATURES BASELINE ==========")

baseline = evaluate_features(current_features)

baseline["Removed Feature"] = None
baseline["Step"] = 0

results.append(baseline)
selection_path.append(baseline)

current_f1 = baseline["Validation F1"]

print(f"Baseline F1: {current_f1:.4f}")

pd.DataFrame(results).to_csv(
    "CNN_Feature_Search.csv", index=False
)

# ============================================
# 5. BACKWARD FEATURE SELECTION
# ============================================

try:

    for step in range(1, len(weather_features) + 1):

        removable_features = [
            f for f in current_features
            if f not in fixed_features
        ]

        if not removable_features:
            break

        print(
            f"\n========== STEP {step} ==========",
            flush=True
        )

        print(
            f"Current features: {len(current_features)}"
        )

        candidates = []

        for i, removed_feature in enumerate(
            removable_features, start=1
        ):

            candidate_features = [
                f for f in current_features
                if f != removed_feature
            ]

            print(
                f"\nCandidate {i}/{len(removable_features)}"
                f" | Removing {removed_feature}",
                flush=True
            )

            result = evaluate_features(
                candidate_features
            )

            result["Removed Feature"] = removed_feature
            result["Step"] = step

            candidates.append(result)
            results.append(result)

            print(
                f"Mean Validation F1: "
                f"{result['Validation F1']:.4f}"
            )

            print(
                f"Mean Validation Accuracy: "
                f"{result['Validation Accuracy']:.4f}"
            )

            # Save after every completed combination
            pd.DataFrame(results).to_csv(
                "CNN_Feature_Search.csv",
                index=False
            )

        # Find best removal
        best_candidate = max(
            candidates,
            key=lambda r: r["Validation F1"]
        )

        candidate_f1 = best_candidate["Validation F1"]

        improvement = candidate_f1 - current_f1

        print(
            f"\nBest removal: "
            f"{best_candidate['Removed Feature']}"
        )

        print(f"New F1: {candidate_f1:.4f}")
        print(f"F1 change: {improvement:+.4f}")

        # Stop if removing features no longer helps
        if STOP_EARLY and improvement <= 0:

            print(
                "\nNo feature removal improved F1."
                "\nStopping backward selection."
            )

            break

        current_features = best_candidate["Features"].copy()
        current_f1 = candidate_f1

        selection_path.append(best_candidate)

        print(
            f"Remaining features: {len(current_features)}"
        )

except KeyboardInterrupt:

    print(
        "\nSearch interrupted. "
        "Completed combinations have been saved."
    )

# ============================================
# 6. RANK ALL FEATURE COMBINATIONS
# ============================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    "Validation F1",
    ascending=False
).reset_index(drop=True)

results_df["Weather Features"] = (
    results_df["Features"].apply(
        lambda features: [
            f for f in features
            if f in weather_features
        ]
    )
)

print("\n========== TOP 10 FEATURE COMBINATIONS ==========")

display(
    results_df[
        [
            "Number of Features",
            "Weather Features",
            "Train F1",
            "Validation F1",
            "Validation Accuracy",
            "F1 Gap"
        ]
    ].head(10).round(4)
)

# ============================================
# 7. BEST FEATURE COMBINATION
# ============================================

best_result = results_df.iloc[0]

best_features = best_result["Features"]

print("\n========== BEST FEATURES ==========")

for feature in best_features:
    print(feature)

print(f"\nNumber of features: {len(best_features)}")

print(
    f"Best Validation F1: "
    f"{best_result['Validation F1']:.4f}"
)

print(
    f"Best Validation Accuracy: "
    f"{best_result['Validation Accuracy']:.4f}"
)

print(
    f"F1 improvement over baseline: "
    f"{best_result['Validation F1'] - baseline['Validation F1']:+.4f}"
)

print(
    f"Total runtime: "
    f"{(time.time() - total_start) / 60:.2f} minutes"
)

# ============================================
# 8. BEST COMBINATIONS BY ACCURACY
# ============================================

print("\n========== TOP 5 BY ACCURACY ==========")

display(
    results_df.sort_values(
        "Validation Accuracy",
        ascending=False
    ).head(5).round(4)
)

# ============================================
# 9. F1 VS NUMBER OF FEATURES
# ============================================

path_df = pd.DataFrame(selection_path)

plt.figure(figsize=(10, 5))

plt.plot(
    path_df["Number of Features"],
    path_df["Train F1"],
    marker="o",
    label="Training F1"
)

plt.plot(
    path_df["Number of Features"],
    path_df["Validation F1"],
    marker="o",
    label="Validation F1"
)

plt.gca().invert_xaxis()

plt.xlabel("Number of Features")
plt.ylabel("F1 Score")
plt.title("CNN Backward Feature Selection")
plt.legend()
plt.grid(True)
plt.show()

# ============================================
# 10. TOP 10 FEATURE COMBINATIONS
# ============================================

top_10 = results_df.head(10)

positions = np.arange(len(top_10))

plt.figure(figsize=(12, 6))

plt.bar(
    positions - 0.2,
    top_10["Train F1"],
    width=0.4,
    label="Training F1"
)

plt.bar(
    positions + 0.2,
    top_10["Validation F1"],
    width=0.4,
    label="Validation F1"
)

plt.xticks(
    positions,
    [
        f"Rank {i+1}\n{n} features"
        for i, n in enumerate(
            top_10["Number of Features"]
        )
    ]
)

plt.ylabel("F1 Score")
plt.title("Top 10 CNN Feature Combinations")
plt.ylim(0, 1)
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()


# Save final ranked results
results_df.to_csv("CNN_Final_Results.csv", index=False)

# Save best feature combination
pd.DataFrame({"Feature": best_features}).to_csv(
    "CNN_Best_Features.csv", index=False
)

print("All results saved successfully!", flush=True)
```
